**Objective:**

```python
from scapy.all import IP, TCP, UDP, Raw, wrpcap, rdpcap
```

```python
packets_to_save = [
    IP(src="192.168.1.10", dst="192.168.1.20") /
    TCP(sport=1234, dport=80) /
    Raw(load="HTTP Request"),

    IP(src="192.168.1.20", dst="192.168.1.10") /
    UDP(sport=53, dport=5353) /
    Raw(load="DNS Response"),

    IP(src="192.168.1.10", dst="8.8.8.8") /
    TCP(sport=5432, dport=443) /
    Raw(load="HTTPS Data")
]

wrpcap("sample_packets.pcap", packets_to_save)

print("Sample packet capture created successfully!")
```

```python
packets = rdpcap("sample_packets.pcap")

print("Total packets:", len(packets))
```

```python
for i, packet in enumerate(packets, 1):
    print(f"\n--- Packet {i} ---")

    if IP in packet:
        print("Source IP      :", packet[IP].src)
        print("Destination IP :", packet[IP].dst)

    if TCP in packet:
        print("Protocol       : TCP")
    elif UDP in packet:
        print("Protocol       : UDP")
    else:
        print("Protocol       : Other")

    if Raw in packet:
        print("Payload        :", packet[Raw].load.decode(errors="replace"))
    else:
        print("Payload        : No payload")
```

```python
print("\n===== Network Packet Analyzer Summary =====")
print("Total packets analyzed:", len(packets))
print("Packet analysis completed successfully.")
```
